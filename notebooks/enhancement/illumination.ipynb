### Illumination

```python
import cv2
import os
import numpy as np
import matplotlib.pyplot as plt

input_folder = "../frames/enhanced/2_NoiseReduced"
output_folder = "../frames/enhanced/3_IlluminationCorrected"

os.makedirs(output_folder, exist_ok=True)

max_images = 3
count = 0

# Contrast Stretching function
def contrast_stretch(img):
    min_val = np.min(img)
    max_val = np.max(img)
    stretched = (img - min_val) * (255 / (max_val - min_val))
    return stretched.astype(np.uint8)

for file in os.listdir(input_folder):
    if file.endswith((".jpg", ".png", ".jpeg")):

        path = os.path.join(input_folder, file)
        img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)

        if img is None:
            continue

        # Contrast Stretching
        contrast_img = contrast_stretch(img)

        # Histogram Equalization
        hist_eq_img = cv2.equalizeHist(img)

        # Save one (final choice → histogram equalized)
        cv2.imwrite(os.path.join(output_folder, file), hist_eq_img)

        # Visual Analysis (Before and After)
        if count < max_images:

            plt.figure(figsize=(12,4))

            plt.subplot(1,3,1)
            plt.imshow(img, cmap="gray")
            plt.title("Original")
            plt.axis("off")

            plt.subplot(1,3,2)
            plt.imshow(contrast_img, cmap="gray")
            plt.title("Contrast Stretching")
            plt.axis("off")

            plt.subplot(1,3,3)
            plt.imshow(hist_eq_img, cmap="gray")
            plt.title("Histogram Equalization")
            plt.axis("off")

            plt.suptitle(f"Illumination Correction: {file}")
            plt.show()

            # Histogram Comparison
            plt.figure(figsize=(8,4))

            hist1 = cv2.calcHist([img],[0],None,[256],[0,256])
            hist2 = cv2.calcHist([hist_eq_img],[0],None,[256],[0,256])

            plt.plot(hist1, label="Original")
            plt.plot(hist2, label="Equalized")
            plt.title("Histogram Improvement")
            plt.legend()
            plt.show()

            count += 1

print("Illumination correction completed!")
```
