```python
import cv2
import os
import matplotlib.pyplot as plt

frames_folder = "../frames/enhanced/1_Grayscale"
output_folder = "../frames/enhanced/2_NoiseReduced"

# Process only few images for analysis
max_images = 3
count = 0

for file in os.listdir(frames_folder):
    if not file.endswith((".jpg", ".png", ".jpeg")):
        continue

    path = os.path.join(frames_folder, file)
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)

    if img is None:
        continue

    # NOISE REMOVAL (Median Filter)
    denoised = cv2.medianBlur(img, 5)

    # Save the denoised image
    cv2.imwrite(os.path.join(output_folder, file), denoised)

    # Visual Analysis (Before and After)
    if count < max_images:
        plt.figure(figsize=(10,4))

        plt.subplot(1,2,1)
        plt.imshow(img, cmap="gray")
        plt.title("Original (Noisy)")
        plt.axis("off")

        plt.subplot(1,2,2)
        plt.imshow(denoised, cmap="gray")
        plt.title("Denoised")
        plt.axis("off")

        plt.suptitle(f"Noise Reduction Analysis: {file}")
        plt.show()

        # Histrogram Comparison
        plt.figure(figsize=(6,4))

        hist1 = cv2.calcHist([img],[0],None,[256],[0,256])
        hist2 = cv2.calcHist([denoised],[0],None,[256],[0,256])

        plt.plot(hist1, label="Original")
        plt.plot(hist2, label="Denoised")
        plt.title("Histogram Comparison")
        plt.legend()
        plt.show()

        count += 1

print("Noise analysis + reduction completed.")
```
