```python
import cv2
import os

def extract_video(video_path, output_folder):

    # get video name
    video_name = os.path.basename(video_path).split('.')[0]

    cap = cv2.VideoCapture(video_path)

    # check if video loaded
    if not cap.isOpened():
        print("Error: Cannot open video:", video_path)
        return

    frame_no = 0
    saved_no = 0

    while True:
        ret, frame = cap.read()

        if ret == False:
            break

        # save every 5th frame
        if frame_no % 5 == 0:
            filename = video_name + "_frame_" + str(frame_no) + ".jpg"
            cv2.imwrite(os.path.join(output_folder, filename), frame)
            saved_no += 1

        frame_no = frame_no + 1

    cap.release()

    print("DONE:", video_name, "Saved:", saved_no)
```

```python
video_path = "../videos/alligator_cracking/aligatorCracking_1.mp4"
output_folder = "../frames/extracted/alligator_cracking"
extract_video(video_path, output_folder)

video_path = "../videos/alligator_cracking/aligatorCracking_2.mp4"
output_folder = "../frames/extracted/alligator_cracking"
extract_video(video_path, output_folder)

video_path = "../videos/alligator_cracking/aligatorCracking_3.mp4"
output_folder = "../frames/extracted/alligator_cracking"
extract_video(video_path, output_folder)

video_path = "../videos/alligator_cracking/aligatorCracking_4.mp4"
output_folder = "../frames/extracted/alligator_cracking"
extract_video(video_path, output_folder)

video_path = "../videos/alligator_cracking/aligatorCracking_5.mp4"
output_folder = "../frames/extracted/alligator_cracking"
extract_video(video_path, output_folder)
```
