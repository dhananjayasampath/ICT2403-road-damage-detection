### Extract Frames

```python
import cv2
import os

def extract_frames(video_path, output_folder):
    cap = cv2.VideoCapture(video_path)

    # Get video name
    video_name = os.path.basename(video_path).split(".")[0]

    frame_no = 0
    saved_no = 0

    # Check if video loaded
    if not cap.isOpened():
        print("Error: Could not open video.")
    else:
        while True:
            # Capture frame-by-frame
            ret, frame = cap.read()
            
            if not ret:
                break

            # Save every 5th frame
            if frame_no %5 == 0:  
                filename = video_name + "_frame_" + str(frame_no) + ".jpg"
                cv2.imwrite(os.path.join(output_folder, filename), frame)
                saved_no += 1
            frame_no += 1

        cap.release()

        print("Done:",video_path, "Saved:", saved_no)


```

```python
# Alligator Cracking videos
# List of video paths
video_paths = [
    "../videos/alligatorCracking/aligatorCracking_1.mp4",
    "../videos/alligatorCracking/aligatorCracking_2.mp4",
    "../videos/alligatorCracking/aligatorCracking_3.mp4",
    "../videos/alligatorCracking/aligatorCracking_4.mp4"
]

# Output folder
output_folder = "../frames/extracted/alligatorCracking"

# Create folder
if not os.path.exists(output_folder):
    os.makedirs(output_folder)

# Loop through videos
for video_path in video_paths:
    extract_frames(video_path, output_folder)
```
