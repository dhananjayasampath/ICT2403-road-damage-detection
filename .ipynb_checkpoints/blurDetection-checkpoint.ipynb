```python
import cv2
import numpy as np
import os
import matplotlib.pyplot as plt
import pandas as pd

# -----------------------------------------
# Input Folder
# -----------------------------------------
input_folder = "frames"

frame_files = sorted(os.listdir(input_folder))

sharpness_values = []
valid_frames = []

print("\nCalculating Initial Sharpness (Laplacian Variance)...\n")

# -----------------------------------------
# STEP 1: Compute Initial Sharpness
# -----------------------------------------
for file in frame_files:

    if not file.lower().endswith((".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff")):
        continue

    path = os.path.join(input_folder, file)
    image = cv2.imread(path)

    if image is None:
        continue

    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    lap_var = cv2.Laplacian(gray, cv2.CV_64F).var()

    sharpness_values.append(lap_var)
    valid_frames.append((file, gray, lap_var))

print("Total Valid Frames:", len(valid_frames))

if len(valid_frames) == 0:
    print("No valid images found.")
    exit()

# -----------------------------------------
# STEP 2: Adaptive Blur Threshold
# -----------------------------------------
mean_sharpness = np.mean(sharpness_values)
std_sharpness = np.std(sharpness_values)
threshold = mean_sharpness - std_sharpness

print("\nMean Sharpness:", round(mean_sharpness,2))
print("Adaptive Blur Threshold:", round(threshold,2))

# -----------------------------------------
# STEP 3: Blur Reduction + Analysis
# -----------------------------------------
results = []
display_limit = 5
display_counter = 0

for file, gray, sharp_before in valid_frames:

    if sharp_before < threshold:

        # Apply Laplacian Sharpening
        laplacian = cv2.Laplacian(gray, cv2.CV_64F)
        sharpened = gray - laplacian
        sharpened = np.clip(sharpened, 0, 255).astype(np.uint8)

        sharp_after = cv2.Laplacian(sharpened, cv2.CV_64F).var()

        improvement = sharp_after - sharp_before
        improvement_percent = (improvement / sharp_before) * 100 if sharp_before != 0 else 0

        results.append([
            file,
            sharp_before,
            sharp_after,
            improvement,
            improvement_percent
        ])

        # -----------------------------------------
        # Display Before vs After (First 5 Only)
        # -----------------------------------------
        if display_counter < display_limit:

            plt.figure(figsize=(10,4))

            plt.subplot(1,2,1)
            plt.imshow(gray, cmap='gray')
            plt.title(f"Before\nSharpness: {round(sharp_before,2)}")
            plt.axis("off")

            plt.subplot(1,2,2)
            plt.imshow(sharpened, cmap='gray')
            plt.title(f"After\nSharpness: {round(sharp_after,2)}")
            plt.axis("off")

            plt.suptitle(file)
            plt.tight_layout()
            plt.show()

            display_counter += 1

# -----------------------------------------
# RESULT TABLE
# -----------------------------------------
df = pd.DataFrame(results, columns=[
    "Frame",
    "Sharpness Before",
    "Sharpness After",
    "Improvement",
    "Improvement (%)"
])

print("\n========== BLUR REDUCTION ANALYSIS ==========")
print(df)

# -----------------------------------------
# STATISTICAL SUMMARY
# -----------------------------------------
if not df.empty:

    avg_before = df["Sharpness Before"].mean()
    avg_after = df["Sharpness After"].mean()
    avg_improve = df["Improvement"].mean()
    avg_percent = df["Improvement (%)"].mean()

    print("\nAverage Sharpness Before :", round(avg_before,2))
    print("Average Sharpness After  :", round(avg_after,2))
    print("Average Improvement      :", round(avg_improve,2))
    print("Average Improvement (%)  :", round(avg_percent,2), "%")

    # -----------------------------------------
    # Sharpness Comparison Graph
    # -----------------------------------------
    plt.figure(figsize=(8,5))
    plt.plot(df["Sharpness Before"], label="Before")
    plt.plot(df["Sharpness After"], label="After")
    plt.title("Sharpness Before vs After")
    plt.xlabel("Blurred Frame Index")
    plt.ylabel("Laplacian Variance")
    plt.legend()
    plt.show()

else:
    print("\nNo blurred frames detected.")

print("\nBlur Reduction Analysis Completed Successfully.")
```
